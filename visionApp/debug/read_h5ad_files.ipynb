```python
import hashlib

# Function to calculate the checksum of a file
def calculate_checksum(file_path):
    with open(file_path, 'rb') as file:
        data = file.read()
        checksum = hashlib.md5(data).hexdigest()
    return checksum

# File paths
file_path1 = r"C:\Users\ghari\Documents\OPS\Test_folders\isotope2\tracer_1_Normal_subset_NIA_formatted_backup.csv"
file_path2 = r"C:\Users\ghari\Documents\OPS\Test_folders\isotope2\tracer_1_Normal_subset_NIA_formatted.csv"

# Calculate checksums
checksum1 = calculate_checksum(file_path1)
checksum2 = calculate_checksum(file_path2)

# Check if the checksums are the same
if checksum1 == checksum2:
    print("The checksums are the same.")
else:
    print("The checksums are different.")


# with open(r"C:\Users\ghari\Documents\OPS\Test_folders\isotope2\tracer_1_Normal_subset_NIA_formatted.csv", 'rb') as file:
#     arr = pickle.load(file)

# with open(r"C:\Users\ghari\Documents\OPS\Test_folders\isotope2\tracer_1_Normal_subset_NIA_formatted_backup.csv", 'rb') as file:
#     arr2 = pickle.load(file)
```
